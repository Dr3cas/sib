## Exercise 6: Stratified splitting
Testing `stratified_train_test_split` with the iris dataset and evaluating a `KNNClassifier`.

```python
import numpy as np

from si.io.csv_file import read_csv
from si.model_selection.split import train_test_split, stratified_train_test_split
from si.models.knn_classifier import KNNClassifier

iris = read_csv('../datasets/iris/iris.csv', features=True, label=True)
```

```python
train, test = stratified_train_test_split(iris, test_size=0.2, random_state=42)
print('train:', train.shape(), '| test:', test.shape())
print('train classes:', dict(zip(*np.unique(train.y, return_counts=True))))
print('test classes: ', dict(zip(*np.unique(test.y, return_counts=True))))
```

Comparison with the non-stratified split

```python
train_r, test_r = train_test_split(iris, test_size=0.2, random_state=42)
print('test classes (random split):', dict(zip(*np.unique(test_r.y, return_counts=True))))
```

KNNClassifier: fit -> predict -> score

```python
knn = KNNClassifier(k=3)
knn.fit(train)
predictions = knn.predict(test)
print(predictions[:5])
print('accuracy:', knn.score(test))
```
