## Exercise 1: NumPy array Indexing/Slicing
Examples with the iris dataset, plus examples of `dropna`, `fillna` and `remove_by_index` (Exercise 2).

### 1.1) Load `iris.csv` with the package functions

```python
from copy import copy

import numpy as np

from si.io.csv_file import read_csv

iris = read_csv('../datasets/iris/iris.csv', sep=',', features=True, label=True)
print(iris.shape())
iris.features
```

### 1.2) Penultimate independent variable

```python
penultimate = iris.X[:, -2]
print('Feature:', iris.features[-2])
print('Dimension of the resulting array:', penultimate.shape)
```

### 1.3) Last 10 samples and their mean (per feature)

```python
last_10 = iris.X[-10:]
print(last_10.shape)
print(dict(zip(iris.features, last_10.mean(axis=0))))
```

### 1.4) Samples with values <= 6 in ALL independent variables

```python
mask = (iris.X <= 6).all(axis=1)
selected = iris.X[mask]
print('Number of samples:', selected.shape[0])
```

### 1.5) Samples with a class different from 'Iris-setosa'

```python
not_setosa = iris.X[iris.y != 'Iris-setosa']
print('Number of samples:', not_setosa.shape[0])
```

### Exercise 2: `dropna`, `fillna` and `remove_by_index`

```python
iris_nas = read_csv('../datasets/iris/iris_missing_data.csv', sep=',', features=True, label=True)
print('Original shape:', iris_nas.shape())
print('Number of NaN:', int(np.isnan(iris_nas.X).sum()))
```

**dropna** - removes the samples with at least one NaN (and the corresponding entries of y)

```python
dropped = copy(iris_nas).dropna()
print('Shape after dropna:', dropped.shape(), '| y size:', len(dropped.y))
print('NaN left:', int(np.isnan(dropped.X).sum()))
```

**fillna** - replaces the NaN by a value, the mean or the median of each feature

```python
for value in (0, 'mean', 'median'):
    filled = copy(iris_nas).fillna(value)
    print(f'fillna({value!r}): NaN left =', int(np.isnan(filled.X).sum()), '| shape =', filled.shape())
```

**remove_by_index** - removes a sample by its index (and the corresponding entry of y)

```python
removed = copy(iris_nas).fillna('mean').remove_by_index(0)
print('Shape after remove_by_index(0):', removed.shape(), '| y size:', len(removed.y))
```
