## Exercise 3: SelectPercentile
Testing the `SelectPercentile` class with the iris dataset (classification).

```python
from si.io.csv_file import read_csv
from si.feature_selection.select_percentile import SelectPercentile
from si.statistics.f_classification import f_classification

iris = read_csv('../datasets/iris/iris.csv', features=True, label=True)
print(iris.shape(), iris.features)
```

```python
select = SelectPercentile(score_func=f_classification, percentile=50)
select.fit(iris)
print('F:', select.F)
print('p:', select.p)
```

```python
transformed = select.transform(iris)
print(transformed.shape(), transformed.features)
```

```python
for percentile in (25, 50, 75, 100):
    t = SelectPercentile(percentile=percentile).fit_transform(iris)
    print(percentile, '->', t.features)
```
