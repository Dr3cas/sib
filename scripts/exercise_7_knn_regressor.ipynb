## Exercise 7: KNNRegressor with RMSE
Testing the `KNNRegressor` class with the `cpu.csv` dataset (regression).

```python
from si.io.csv_file import read_csv
from si.model_selection.split import train_test_split
from si.models.knn_regressor import KNNRegressor
from si.metrics.rmse import rmse

cpu = read_csv('../datasets/cpu/cpu.csv', features=True, label=True)
print(cpu.shape(), cpu.features, cpu.label)
```

```python
train, test = train_test_split(cpu, test_size=0.2, random_state=42)
print('train:', train.shape(), '| test:', test.shape())
```

```python
knn = KNNRegressor(k=3)
knn.fit(train)
predictions = knn.predict(test)
print(predictions[:5])
print(test.y[:5])
```

```python
print('rmse (score):', knn.score(test))
print('rmse (function):', rmse(test.y, predictions))
```
